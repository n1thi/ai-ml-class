```python
#Q1. Phone Contact Book

contacts = {}

while True:
    print("\n1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")

    choice = input("Choose an option (1-4): ")

    if choice == "1":
        name = input("Enter name: ").strip()

        if name in contacts:
            print("Contact already exists")
        else:
            phone = input("Enter phone number: ")
            contacts[name] = phone
            print("Contact added")

    elif choice == "2":
        name = input("Enter name to search: ").strip()

        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts saved")
        else:
            for name, phone in contacts.items():
                print(name, ":", phone)

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid option")
```

```python
#Q3. Student Marks Dictionary

marks = {"Aman": 82, "Priya": 91, "Rahul": 45,
         "Sneha": 67, "Kiran": 30}

passed_students = {}
topper = ""
highest_marks = -1

for name, score in marks.items():
    if score >= 40:
        print(name, "- Pass")
        passed_students[name] = score
    else:
        print(name, "- Fail")

    if score > highest_marks:
        highest_marks = score
        topper = name

print("Topper:", topper, "-", highest_marks)
print("Passed students:", passed_students)
```

```python

riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

print("Mutual friends:", sorted(riya_friends & arjun_friends))

print("Only Riya's friends:", sorted(riya_friends - arjun_friends))

print("All friends:", sorted(riya_friends | arjun_friends))

print("Suggestions for Riya:", sorted(arjun_friends - riya_friends))
```

```python
#Q5. Character Counter

text = input("Enter text: ")
counts = {}

for char in text:
    if char != " ":
        if char in counts:
            counts[char] += 1
        else:
            counts[char] = 1

print("Character counts:", counts)

if len(counts) == 0:
    print("No characters to count")
else:
    most_frequent = ""
    highest_count = 0

    for char, count in counts.items():
        if count > highest_count:
            most_frequent = char
            highest_count = count

    print("Most frequent character:", most_frequent,
          "(" + str(highest_count) + " times)")
```

```python
#Q6. Restaurant Menu Billing

menu = {"dosa": 60, "idli": 40, "vada": 30,
        "coffee": 25, "tea": 15}

total = 0
print("Menu:", menu)

while True:
    item = input("Enter item (or 'bill' to finish): ").strip().lower()

    if item == "bill":
        break
    elif item in menu:
        total += menu[item]
        print("Added:", item, "- Rs.", menu[item])
    else:
        print("Item not available")

if total > 200:
    discount = total * 0.10
    total -= discount
    print("10% discount applied: Rs.", discount)

print("Total bill: Rs.", total)
```

```python
#Q7. City Coordinates (Tuples)

cities = [("Hyderabad", 17.38, 78.48),
          ("Mumbai", 19.07, 72.87),
          ("Chennai", 13.08, 80.27),
          ("Delhi", 28.61, 77.20)]

northern_city = ""
highest_latitude = -90

for city, latitude, longitude in cities:
    print(city, "- Latitude:", latitude, "Longitude:", longitude)

    if latitude > highest_latitude:
        highest_latitude = latitude
        northern_city = city

print("Farthest north:", northern_city, "-", highest_latitude)

```

```python
#Q8. Library Book Tracker

books = {"Python Basics": 3, "Data Science 101": 0,
         "Machine Learning": 2, "SQL Guide": 1}

book_name = input("Enter book name to borrow: ").strip()

if book_name not in books:
    print("Book not found")

elif books[book_name] == 0:
    print("Currently unavailable")

else:
    books[book_name] -= 1
    print("Book issued successfully")

print("Updated books:", books)
```

```python
#Q9. Group Employees by Department

employees = [("Ravi", "IT"), ("Anjali", "HR"),
             ("Suresh", "IT"), ("Meena", "Finance"),
             ("Arun", "HR"), ("Divya", "IT")]

departments = {}

for name, department in employees:
    if department not in departments:
        departments[department] = []

    departments[department].append(name)

print("Employees by department:", departments)

for department, names in departments.items():
    print(department, "has", len(names), "employees")
```

```python
#Q10. Quiz Answer Checker

answer_key = ("B", "C", "A", "D", "B")

wrong_questions = set()
score = 0

for i in range(len(answer_key)):
    answer = input("Answer for question " + str(i + 1) + ": ").strip().upper()

    if answer == answer_key[i]:
        print("Question", i + 1, "- Correct")
        score += 1
    else:
        print("Question", i + 1, "- Wrong")
        wrong_questions.add(i + 1)

print("Wrong question numbers:", sorted(wrong_questions))
print("Final score:", score, "/ 5")

if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Needs improvement")
```
