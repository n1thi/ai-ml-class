```python
#Q1. Class Marks Report

marks = [78, 45, 92, 33, 67, 88, 51, 29]

passed = 0

for mark in marks:
    if mark >= 35:
        passed += 1

average = sum(marks) / len(marks)

print("Highest marks:", max(marks))
print("Lowest marks:", min(marks))
print("Class average:", average)
print("Students passed:", passed)
```

```python
#Q2. Grocery Shopping List

shopping_list = []

while True:
    item = input("Enter item: ").strip().lower()

    if item == "done":
        break
    elif item == "":
        print("Enter a valid item")
    elif item in shopping_list:
        print("Already added")
    else:
        shopping_list.append(item)

for i in range(len(shopping_list)):
    print(str(i + 1) + ".", shopping_list[i])

print("Total items:", len(shopping_list))
```

```python
#Q3. Username Validator

username = input("Enter username: ")

if len(username) < 5 or len(username) > 12:
    print("Invalid: Username must be 5 to 12 characters long")
elif " " in username:
    print("Invalid: Username cannot contain spaces")
elif username[0].isdigit():
    print("Invalid: Username cannot start with a digit")
else:
    print("Valid username")
```

```python
#Q4. Vowel Counter for a Tweet

sentence = input("Enter a sentence: ").lower()

vowels = 0
consonants = 0
spaces = 0

for char in sentence:
    if char in "aeiou":
        vowels += 1
    elif "a" <= char <= "z":
        consonants += 1
    elif char == " ":
        spaces += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
```

```python
#Q5. Palindrome Name Checker

name = input("Enter your name: ").strip().lower()

if name == name[::-1]:
    print("Bonus! Your name is a palindrome.")
else:
    print("Your name is not a palindrome.")
```

```python
#Q6. Temperature Tracker

temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

hottest_temp = temps[0]
hottest_day = days[0]
fahrenheit = []

for i in range(len(temps)):
    if temps[i] > 37:
        print("Heat alert:", days[i], "-", temps[i], "C")

    if temps[i] > hottest_temp:
        hottest_temp = temps[i]
        hottest_day = days[i]

    fahrenheit.append(temps[i] * 9 / 5 + 32)

print("Hottest day:", hottest_day, "-", hottest_temp, "C")
print("Temperatures in Fahrenheit:", fahrenheit)
```

```python
#Q7. Word Frequency in a Sentence

sentence = input("Enter a sentence: ")
words = sentence.split()

if len(words) == 0:
    print("No words entered")
else:
    longest = words[0]
    shortest = words[0]
    capital_count = 0

    for word in words:
        if len(word) > len(longest):
            longest = word
        if len(word) < len(shortest):
            shortest = word
        if word[0].isupper():
            capital_count += 1

    print("Longest word:", longest)
    print("Shortest word:", shortest)
    print("Words starting with capital letter:", capital_count)
```

```python
#Q8. Remove Negative Balances

balances = [1500, -200, 3000, -50, 0, 750, -1000]

positive_accounts = []
negative_accounts = []
total_owed = 0

for balance in balances:
    if balance > 0:
        positive_accounts.append(balance)
    elif balance < 0:
        negative_accounts.append(balance)
        total_owed += -balance

print("Positive accounts:", positive_accounts)
print("Negative accounts:", negative_accounts)
print("Total amount owed:", total_owed)
```

```python
#Q9. Password Strength Checker

password = input("Enter password: ")

has_upper = False
has_lower = False
has_digit = False

for char in password:
    if char.isupper():
        has_upper = True
    if char.islower():
        has_lower = True
    if char.isdigit():
        has_digit = True

conditions_met = 0

if has_upper:
    conditions_met += 1
if has_lower:
    conditions_met += 1
if has_digit:
    conditions_met += 1
if len(password) >= 8:
    conditions_met += 1

if conditions_met == 4:
    print("Strong")
elif conditions_met >= 2:
    print("Medium")
else:
    print("Weak")
```

```python
#Q10. Simple Name Initials Generator

full_name = input("Enter full name: ").strip()
parts = full_name.split()

if len(parts) == 0:
    print("Please enter a name")
else:
    initials = ""
    short_name = ""

    for part in parts:
        initials += part[0].upper() + "."

    for i in range(len(parts) - 1):
        short_name += parts[i][0].upper() + ". "

    short_name += parts[-1].capitalize()

    print("Initials:", initials)
    print("Title case:", full_name.title())
    print("Short name:", short_name)
```
